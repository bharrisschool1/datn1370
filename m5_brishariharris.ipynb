```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/papayuh/datn1370-datasets/main/data/synthetic/student_budgets.csv")

df.head()
```

```python
import matplotlib.pyplot as plt

avg_expenses = df[['rent_usd' , 'food_usd' , 'transport_usd' , 'entertainment_usd']].mean()

avg_expenses.plot(kind='bar')

plt.title('Average Student Expenses')
plt.xlabel('Expense Category')
plt.ylabel('Average Amount (USD)')
plt.show()
```

This bar chart shows the average amount students spend in different expense categories. Rent has the highest average expense, followed by food. Transportation and entertainment have lower average cost. A bar chart is useful because it makes it easy to compare spending across different categories.

```python
plt.hist(df['monthly_income_usd'], bins=10)

plt.title('Distribution of monthly Student Income')
plt.xlabel('Monthly Income (USD)')
plt.ylabel('Number of Students')

plt.show()
```

The histogram shows monthly income is distributed among students. Most students appear to earn income within a certain range, while fewer students are at the highest and lowest income levels. A hisogram is useful because it shows the distribution and frequency of numerical data.
